```python
import json

purchases = {}

with open(r"C:\Users\Artem\Downloads\purchase_log.txt", encoding='utf-8') as file:
    for line in file:
        data = json.loads(line)
        purchases[data['user_id']] = data['category']

for user_id, category in list(purchases.items())[:50]:
    print(str(user_id) + " '" + str(category) + "'")
```

```python
import csv

with open(r"C:\Users\Artem\Downloads\visit_log.csv", encoding='utf-8') as visit_file, \
     open(r"C:\Users\Artem\Downloads\funnel.csv", 'w', encoding='cp1251', newline='') as funnel_file:

    reader = csv.reader(visit_file)
    writer = csv.writer(funnel_file)

    header = next(reader)
    writer.writerow(['user_id', 'source', 'category'])

    for row in reader:
        user_id, source = row

        if user_id in purchases:
            writer.writerow([user_id, source, purchases[user_id]])
```
