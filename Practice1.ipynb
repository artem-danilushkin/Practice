```python
number = 123123

ticket = str(number)

sum_first = int(ticket[0]) + int(ticket[1]) + int(ticket[2])
sum_last = int(ticket[3]) + int(ticket[4]) + int(ticket[5])

if sum_first == sum_last:
    print("Счастливый билет")
else:
    print("Несчастливый билет")
```
