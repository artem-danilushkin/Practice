```python
from datetime import datetime

class Account:
    def __init__(self, name, starting_balance=0):
        self.name = name
        self.balance = starting_balance
        self.history = []  

        self._add_history("Аккаунт создан", starting_balance)

    def _add_history(self, operation_type, amount):
        #Добавляет запись в историю операций
        self.history.append({
            "timestamp": datetime.now(),
            "operation": operation_type,
            "amount": amount,
            "balance": self.balance
        })

    def deposit(self, amount):
        #Положить деньги на счёт
        if amount <= 0:
            raise ValueError("Сумма для депозита должна быть положительной")
        self.balance += amount
        self._add_history("Депозит", amount)

    def withdraw(self, amount):
        #Снять деньги со счета
        if amount <= 0:
            raise ValueError("Сумма для снятия должна быть положительной")
        if amount > self.balance:
            raise ValueError("Недостаточно средств на счете")
        self.balance -= amount
        self._add_history("Снятие", amount)

    def show_history(self):
        #Показать историю операций
        for record in self.history:
            time_str = record["timestamp"].strftime("%Y-%m-%d %H:%M:%S")
            print(f"{time_str} | {record['operation']:10} | Amount: {record['amount']:8} | Balance: {record['balance']:8}")

    def __str__(self):
        return f"Account({self.name}, Balance: {self.balance})"
```
