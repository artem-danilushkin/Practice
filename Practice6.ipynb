```python
from datetime import datetime

formats = [
    "%A, %B %d, %Y", 
    "%A, %d.%m.%y",   
    "%A, %d %B %Y"     
]

print("Введите дату (или q для выхода):")

while True:
    date_str = input()

    if date_str.lower() == 'q':
        print("Работа программы завершена.")
        break

    parsed = False

    for fmt in formats:
        try:
            date_obj = datetime.strptime(date_str, fmt)
            print(date_obj)
            parsed = True
            break
        except ValueError:
            continue
            
    if not parsed:
        print("Неверный формат! Введите корректную дату!")
        continue
```
