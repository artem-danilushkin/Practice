```python
queries = [
    'смотреть сериалы онлайн',
    'новости спорта',
    'афиша кино',
    'курс доллара',
    'сериалы этим летом',
    'курс по питону',
    'сериалы про спорт',
]

total = len(queries)
stats = {}

for query in queries:
    words_count = len(query.split())
    stats[words_count] = stats.get(words_count, 0) + 1

for words, count in sorted(stats.items()):
    percent = round(count / total * 100, 2)
    print(f'Поисковых запросов, содержащих {words} слов(а): {percent}%')
```
